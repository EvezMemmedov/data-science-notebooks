```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data1 = np.linspace(0, 10, 20)
```

```python
data2 = data1 ** 2
```

```python
my_fig, my_axes = plt.subplots()
my_axes.plot(data1, data2)
my_axes.plot(data2, data1)
plt.show()
```

```python
(new_fig, new_axes) = plt.subplots()
new_axes.plot(data1, data1 + 2, color = "blue", linewidth = 1)
new_axes.plot(data1, data1 + 3, color = "red", linewidth = 4)
new_axes.plot(data1, data1 + 4, color = "red", linestyle = "-.")
new_axes.plot(data1, data1 + 5, color = "black", linestyle = "--", marker = "o", markerfacecolor = "yellow", markersize = 10)


plt.show()
```

```python
plt.scatter(data1, data2, color = "red")
plt.show()
```
