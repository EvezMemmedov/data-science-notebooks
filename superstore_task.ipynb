```python
#13. Which category has the highest total sales? Compare the categories with a bar chart.
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales.plot(kind='bar', title='total sales by category', xlabel='Category', ylabel='total sales');
```

```python
#14. Which region has the highest total profit? Compare the regions with a bar chart.
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
region_profit.plot(kind='bar', title = 'profit by region', xlabel='region', ylabel='profit');
```

```python
#15. Are small or large sales amounts more common? Draw a histogram using 20 bins and describe its shape.
df['Sales'].plot(kind='hist', bins=20, title='Distribution of Sales Amounts', xlabel='Sales Amount')
```
