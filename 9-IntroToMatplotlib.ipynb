```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
age_list = [12, 34, 45, 46, 50, 59, 65]
```

```python
weight_list = [40, 60, 70, 80, 90, 100, 110]
```

```python
plt.plot(age_list, weight_list, "r--")
plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Age vs Weight")
plt.show()
```

```python
numpy_arr1 = np.linspace(0, 10, 20)
```

```python
numpy_arr1
```

```python
numpy_arr2 = numpy_arr1 ** 3
```

```python
plt.plot(numpy_arr1, numpy_arr2, "g*-")
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.plot(numpy_arr1, numpy_arr2, "g*-")
plt.subplot(1, 2, 2)
plt.plot(numpy_arr2, numpy_arr1, "r--")
plt.show()
```

```python
my_figure = plt.figure()
figure_axis = my_figure.add_axes([0.1, 0.1, 0.7, 0.7])
figure_axis.plot(numpy_arr1, numpy_arr2, "g-")
figure_axis.set_xlabel("X-axis")
figure_axis.set_ylabel("Y-axis")
figure_axis.set_title("My Figure Title")
plt.show()
```

```python
new_fig = plt.figure(dpi = 100)
new_axis = new_fig.add_axes([0.1, 0.1, 0.7, 0.7])
new_axis.plot(numpy_arr1, numpy_arr2 ** 2, label = "y = x^2")
new_axis.plot(numpy_arr1, numpy_arr2 ** 3, label = "y = x^3")
new_axis.legend(loc=6)
plt.show()
```
